```
import numpy as np
import matplotlib.pyplot as plt

def binario_a_imagen(archivo_binario, tamaño=256):
    with open(archivo_binario, 'rb') as f:
        bytes = np.frombuffer(f.read(), dtype=np.uint8)

    # Ajustar tamaño
    bytes = bytes[:tamaño*tamaño]
    if len(bytes) < tamaño*tamaño:
        bytes = np.pad(bytes, (0, tamaño*tamaño - len(bytes)))

    imagen = bytes.reshape((tamaño, tamaño))
    return imagen

# Ejemplo con un binario (simulado)
imagen_malware = np.random.randint(0, 256, (256, 256))  # Simulación
plt.imshow(imagen_malware, cmap='gray')
plt.title("Imagen de malware (bytes como píxeles)")
plt.show()
```
